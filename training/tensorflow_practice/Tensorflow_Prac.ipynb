```python
import tensorflow as tf
```

```python
daily_sales_number=[21,22,-108,31,-1,32,34,31]
```

```python
tf_dataset=tf.data.Dataset.from_tensor_slices(daily_sales_number)
tf_dataset
```

```python
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
#print 1st 3 elements
for sales in tf_dataset.take(3):
    print(sales.numpy())
```

```python
#filter out negative numbers
tf_2dataset=tf_dataset.filter(lambda x:x>0)
for sales in tf_2dataset.as_numpy_iterator():
    print(sales)
```

```python
#multiply all elements with 72
tf_dataset=tf_2dataset.map(lambda x:x*72)
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
#pass values as batch
for sales_batch in tf_dataset.batch(2):
    print(sales_batch.numpy())
```

```python
#doing all operations as a single command line
tf_dataset = tf.data.Dataset.from_tensor_slices(daily_sales_number)
tf_dataset=tf_dataset.filter(lambda x:x>0).map(lambda y:y*72).shuffle(2).batch(2)
for sales in tf_dataset.as_numpy_iterator():
    print(sales)
```

```python
image_count=len(image_ds)
image_count
```

```python
train_size=int(image_count*0.8)

train_ds=image_ds.take(train_size)
test_ds=image_ds.skip(train_size)
```

```python
def get_label(file_path):
    parts=tf.strings.split(file_path,'\\')
    return parts[-2]
                           
```

```python
def process_image(file_path):
    labels=get_label(file_path)

    img=tf.io.read_file(file_path)
    img=tf.image.decode_jpeg(img)
    img=tf.image.resize(img,[128,128])

    return img,labels
    
```

```python
train_ds=train_ds.map(process_image)
for img,labels in train_ds.take(3):
    print(img)
    print(labels)
```

```python
def scale(img,labels):
    return img/255,labels
```

```python
train_ds=train_ds.map(scale)
for img,labels in train_ds.take(4):
    print(img.numpy()[0][0])
    print(labels)
```
